#### Ejercicio 16

¿Qué puedes decir acerca del precio por metro cuadrado (precio/m2) entre los municipios de 'Getafe' y 'Alcorcón'? (★★☆)

> Debes usar el subset obtenido en la pregunta 13
>
> Pista: Crea una nueva columna llamada `pps` (price per square en inglés) y luego analiza los valores

````python
if "df_sur" not in globals():
    poblaciones_sur = ["Fuenlabrada", "Leganés", "Getafe", "Alcorcón"]
    df_sur = df.loc[df["level5"].isin(poblaciones_sur)].copy()

df_sur["pps"] = df_sur["price"] / df_sur["surface"]
pps_getafe = df_sur.loc[df_sur["level5"].eq("Getafe"), "pps"].mean()
pps_alcorcon = df_sur.loc[df_sur["level5"].eq("Alcorcón"), "pps"].mean()
print(f"Precio por m² promedio en Getafe: {pps_getafe:,.2f} €/m²")
print(f"Precio por m² promedio en Alcorcón: {pps_alcorcon:,.2f} €/m²")
````

En los anuncios de este dataset, el promedio es mayor en Alcorcón (2.239,30 €/m²) que en Getafe (2.066,31 €/m²). La diferencia observada es de unos 172,99 €/m².
